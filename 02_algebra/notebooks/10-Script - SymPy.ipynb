# Inecuaciones

Resuelve la inecuación
$$x^2 +2 < 0$$

````python
from sympy import *
x = Symbol('x')
ineq_obj = x**2 + 2 > 0
conv = ineq_obj.lhs
p = Poly(conv, x)
rel = ineq_obj.rel_op
solve_poly_inequality(p, rel)
````

````python
plot(x**2 + 2)
````

Resuelve la inecuación racional

$$\frac{x-1}{x+2}>0$$

````python
from sympy import *
x = Symbol('x')
ineq_obj = ((x - 1)/(x + 2)) > 0
conv = ineq_obj.lhs
# Como la ecuación es una fracción debemos obtener
# el numerador y denominador
numer, denom = conv.as_numer_denom()
p1 = Poly(numer)
p2 = Poly(denom)
rel = ineq_obj.rel_op

# Solucionar la inecuación
solve_rational_inequalities([[((p1, p2), rel)]])
````
