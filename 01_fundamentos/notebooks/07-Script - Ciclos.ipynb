# Ciclo while

Es una sentencia que se repite mientras se cumpla una condición. Cada una de las repeticiones tiene el nombre de iteraciones.

```python
# Demostración del ciclo while

n = int(input("Ingresa el numero positivo: "))

i = 1
while i <= n:
    print(i)
    i = i + 1
```

```python
# Crea un script que identifique si un número es primo o no

n = int(input("Ingresa un numero entero positivo: "))

EsPrimo = True
i = 2
while i <= n/2:
    if n % i == 0: 
        EsPrimo = False
        break
    i = i + 1

if EsPrimo: print("El numero {} es primo".format(n))
else: print("El numero {} no es primo".format(n))
```

# Ciclo while con la cláusula else

```python
# Crea un script que identifique si un número es primo o no

n = int(input("Ingresa un numero entero positivo: "))

EsPrimo = True
i = 2
while i <= n/2:
    if n % i == 0: 
        EsPrimo = False
        break
    i = i + 1
else: 
    EsPrimo = True

if EsPrimo: print("El numero {} es primo".format(n))
else: print("El numero {} no es primo".format(n))
```

# Ciclo for

## Forma 1 usando range()

```python
# Programa que numere la secuencia del 0 al 5

for i in range(6): 
    print(i, end="")
```

## Forma 2 range(inicio, final)

```python
# Programa que haga la secuencia de los numeros del 4 al 10

for i in range(4, 11):
    print(i)
```

## Forma 3 range(inicio, final, salto)

```python
# Programa que haga la secuencia de los numeros del 4 al 10 de 2 en 2

for i in range(4, 11, 2):
    print(i)
```

# Programa que genera todos los números primos con un límite

```python
# Pedimos el límite al usuario
lim = int(input("Ingresa el límite de números: "))

for n in range(2, lim + 1): # Ciclo para el límite de numeros
    i = 2
    while i <= n / 2:       # Ciclo para deducir si es primo o no
        if n % i == 0:      # Si el residuo es cero paramos el ciclo
            break           # ya que el numero es primo
        i = i + 1
    else:                   # Imprimimos el numero n porque ese es primo
        print(n, end=" ")
```

```python
# Pedimos el límite al usuario
lim = int(input("Ingresa el límite de números: "))

for n in range(2, lim + 1):             # Ciclo para el límite de numeros
    for i in range(2, int(n / 2) + 1):  # Ciclo para deducir si es primo
        if n % i == 0:                  # Condición para saber si es primo
            break
    else:
        print(n)                        # Presentar el numero primo
```

```python
# Programa que suma los digitos de un numero dado

n = int(input("Ingresa un numero entero: "))        # Ingresa el numero
sum = 0                                             # Inicializa la suma

while n > 0:
    digit = n % 10
    sum += digit
    n //= 10

print("La suma de los digitos es:", sum)
```

```python
# Programa que pone los números al revés

n = int(input("Ingresa un numero entero: "))
reversa = 0

while n > 0:                            # Inicializar el ciclo hasta cero
    digit = n % 10                      # Separamos el numero más a la derecha
    reversa = reversa * 10 + digit      # Multiplicar por 10 + el digito
    n //= 10                            # Tomamos el cociente de la división

print("El numero al revés es:", reversa)
```

```python
# Crea un programa que determine si un numero es un palíndromo

# 11 es palíndromo
# 12321 es palíndromo

n = int(input("Ingresa un numero entero: "))
num = n
reversa = 0

while n > 0:
    digit = n % 10
    reversa = reversa * 10 + digit
    n //= 10

if num == reversa: print("El numero {} es un palíndromo".format(num))
else: print("El numero {} no es un palíndromo".format(num))
```
