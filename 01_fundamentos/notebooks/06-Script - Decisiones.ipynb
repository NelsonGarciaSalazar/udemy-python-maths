# La sentencia IF

La sentencia *IF* es una condicional que se ejecuta en un bloque de código en caso de que la condición ingresada se cumpla, teniendo la siguiente estructura:

if *condición*: 
  **Cosas por hacer en caso de que la condición se cumpla**

```python
# Programa que pida el nombre del usuario y su edad
# con el fin de decidir si puede conducir o no

nombre = input("Ingresa tu nombre: ")
edad = int(input("Ingresa tu edad: "))

#if edad >= 18: print("Hola {}, tú puedes conducir".format(nombre))
#if edad < 18: print("Hola {}, tú no puedes conducir".format(nombre))

if edad >= 18:
    print("Hola {}, tú puedes conducir".format(nombre))
else:
    print("Hola {}, tú no puedes conducir".format(nombre))
```

# elif

Este es muy similar al anterior pero, lo que nos permite es hacer condiciones dentro de las condiciones que no se cumplen.

```python
# prorgama que clasifica un numero entero si este es:

# 1. Es positivo
# 2. Es negativo
# 3. Es cero

x = int(input("Ingresar un número entero: "))

if x > 0:
    print("{} es positivo".format(x))
elif x < 0:
    print("{} es negativo".format(x))
else:
    print("Es cero".format(x))
```

```python
# Crear un programa que identifique, a partir de un punto
# en el plano, en qué cuadrante se encuentra.

x = int(input("Ingresa la coordenada en el eje de las x: "))
y = int(input("Ingresa la coordenada en el eje de las y: "))

if x == 0 and y == 0:
    print("El punto está en el origen")
elif y == 0:
    print("El punto está sobre el eje de las x")
elif x == 0:
    print("El punto está sobre el eje de las y")
elif x > 0 and y > 0:
    print("El punto se encuentra en el primer cuadrante")
elif x < 0 and y > 0:
    print("El punto se encuentra en el segundo cuadrante")
elif x < 0 and y < 0:
    print("El punto se encuentra en el tercer cuadrante")
else:
    print("El punto se encuentra en el cuarto cuadrante")
```
