# Imprimir datos

Tanto en programación como en matemáticas es importante poder mostrar la información de la mejor manera posible, por lo que conocer todas las instancias del como mostrar información es importante.

```python
print("Hola mundo!")
print("Hola", "mundo!")
print("La suma de 3 y 5 es:", 3 + 5)
print("Hola", "mundo!", sep = '<->')
print("Hola", "mundo!", "estoy", "usando", "Python", sep = '<->')
print("La suma de 3 y 5 es:", 3 + 5, sep="")
print("Python","es","lo","mejor","para","hacer","matemáticas", end="\n\t en el mundo\n")
print("hola", end="\n\n\n\n")
```
