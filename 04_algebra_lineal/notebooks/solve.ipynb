```python
from sympy import symbols, solve, Eq

def cumple(L, x, y, limite=108):
    total = L + 2 * (x + y)
    return total, total <= limite

# (a) Paquete 1 (todo en pulgadas)
print(cumple(60, 6, 8))      # (88, True)

# (a) Paquete 2: 24 x 24 x 48 pulg, L = la dimensión más larga
print(cumple(48, 24, 24))    # (144, False)

# (b) Longitud máxima con base 9 x 9
L = symbols("L")
print(solve(Eq(L + 2 * (9 + 9), 108), L))   # [72]
```
