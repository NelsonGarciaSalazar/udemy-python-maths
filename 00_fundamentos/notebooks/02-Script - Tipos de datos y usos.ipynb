# Tipos de datos básicos

Una de las primeras cosas que se aprenden de cualquier lenguaje de programación son los tipos de datos que usa. Esto nos puede ayudar a conocer como se representan y que operadores se permiten. Python tiene un lenguaje dinámico que no requiere que una variable sea declarada como un tipo específico, ya que todo dato es considerado un objeto. Los tipos de datos que tiene Python son:

1.   Enteros (```int```)
2.   Números reales (```float```)
3.   Números complejos (```complex```)
4.   Cadenas de caracteres (```str```)
5.   Booleanos (```bool```)

# Variables

Las variables son espacios de memoria que se almacenan dinámicamente en la computadora para poder usar de una manera más sencilla datos que necesitemos en nuestros Scripts. Estos nombres tienen que tener algunas características importantes:

1. Los nombres de las variables no deben contener espacios entre caracteres.
2. No pueden tener caracteres especiales. Mi consejo sería usar solo el **guión bajo**.
3. Los nombres de la variables tienen que ser significativos, es decir, que el nombre represente el tipo de dato que almacenará. No le podemos poner una variable *VacaLola* al número de personas en un parque (si se puede, pero causaría dos o tres malos entendidos).
4. La forma de notación de *camello* o *camell case* es ampliamente usada, y esta depende de la combinación de mayúsuclas y minúsculas, es decir, podemos poner la variable que representa el número de personas en un parque como *PersonasEnParque* y de esta forma se puede interpretar fácilmente que esta variable tendrá almacenada la cantidad de personas en un parque.
5. No usar palabras reservadas del propio lenguaje.

````python
x = 3
y = 3.1415
z = "Esto es una cadena de caracteres"
NumeroDeNiñosEnLaEscuela = 189

print(NumeroDeNiñosEnLaEscuela)

w = 4
````

## Operadores para números complejos

Si tenemos el numero complejo ```3 + 2j``` podemos usar las funciones

1. ```(3 + 2j).real``` Nos regresa la parte real


````python
c = 3 + 2j
c.real
````

2. ```(3 + 2j).imag``` Nos regresa la parte imaginaria



````python
c.imag
````

NOTA: También se puede declarar un numero complejo como ```complex(3,2)``` respetando el primer numero real y el segundo como imaginario.



````python
c = complex(3,2)
c
````

1. Potencia de números complejos ```(a + bj) ** c``` siendo la potencia el número $c$.


````python
c ** 2
````

2. Multiplicación entre números complejos ```(a + bj) * (c + dj)```


````python
d = complex(1,2)
c * d
````

3. División de números complejos ```(a + bj) / (c + dj)```


````python
c / d
````

4. Suma de números complejos ```(a + bj) + (c + dj)```


````python
c + d
````

5. Resta de números complejos ```(a + bj) - (c + dj)```


````python
c - d
````

6. Forma polar de un número complejo ```cmath.polar(a + bj)```


````python
import cmath
cmath.polar(c)
````

````python
cmath.rect(3.605551275463989, 0.5880026035475675)
````

8. Magnitud de un numero complejo ```abs(a + bj)```


````python
abs(c)
````

9. La fase o ángulo del número complejo ```cmath.phase(a + bj)```


````python
cmath.phase(c)
````
