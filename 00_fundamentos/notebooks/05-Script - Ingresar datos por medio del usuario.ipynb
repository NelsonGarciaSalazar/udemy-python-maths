# Introducir datos a python para poderlos manipular

```python
# Hacer un programa que reciba dos datos del usuario y que estos numeros se 
# sumen, resten, multipliquen y dividan.

a = float(input("Da un numero cualquiera: "))
b = float(input("Da otro numero cualquiera: "))
print("La suma de",a,"y",b,"es:",a + b)
print("La resta de",a,"y",b,"es:",a - b)
print("La multiplicación de",a,"y",b,"es:",a * b)
print("La división de",a,"y",b,"es:",a / b)
```

```python
# Programa que pide el nombre y la edad del usuario
# e imprime esta información como un saludo.

nombre = input("Ingresa tu nombre:")
edad = int(input("Ingresa tu edad:"))
peso = float(input("Ingresa tu peso:"))

print("Hola {}, tu tienes {:d} años y pesas {:f}".format(nombre, edad, peso))
```

```python
# Programa que hace operaciones básicas con enteros

import math

x = int(input("Ingresa un numero entero: "))

print("x = {}".format(x))
print("x está entre {} y {}".format(x - 1, x + 1))
print("El cuadrado de x es: {}, y su cubo es: {}".format(x**2, pow(x, 3)))
print("La raíz cuadrada del numero es: {}".format(math.sqrt(x)))
print("El factorial del numero es: {}".format(math.factorial(x)))
```

```python
# Logaritmos y antilogaritmos

import math

x = float(input("Introducir un número real: "))

print("Logaritmo natural y anti-logaritmo")
print("ln(x) = {}, exp(ln(x)) = {}".format(math.log(x), math.exp(math.log(x))))

print("Logaritmo neperiano y su anti-logaritmo")
print("log(x) = {}, anti-logaritmo(log(x)) = {}".format(math.log10(x),
                                                        pow(10, math.log10(x))))
```

```python
# Funciones trigonométricas

angulo = float(input("Ingresa un ángulo en grados: "))

angulo = math.radians(angulo) # Convertimos la variable a radianes

print("sen(x) = ", math.sin(angulo))
print("cos(x) = ", math.cos(angulo))
print("tan(x) = ", math.tan(angulo))
print("cosec(x) = ", 1 / math.sin(angulo))
print("sec(x) = ", 1 / math.cos(angulo))
print("cot(x) = ", 1 / math.tan(angulo))
```

```python
# Demostración de operaciones con números complejos

import cmath

real = float(input("Ingresa la parte real del numero complejo: "))
imag = float(input("Ingresa la parte imaginaria del numero complejo: "))

c = complex(real, imag) # Hacemos nuestro numero complejo

print("El numero complejo es:", c)
print("\tLa parte real es:", c.real)
print("\tLa parte imaginaria es:", c.imag)
print("Su magnitud es:", abs(c))
print("Su ángulo es:", math.degrees(cmath.phase(c)))
```

```python
# Programa que encuentra el interes simple y
# su cantidad

p = float(input("Ingresa el capital a invertir:"))
r = float(input("Ingresa el porcentaje de interés anual:"))
t = float(input("Ingresa la cantidad de tiempo de la inversión:"))

g = (p * r * t) / 100
ct = p + g

print("Interes simple es =", g)
print("La cantidad total = ", ct)
```

$g = \frac{p \cdot r \cdot t}{100}$

$ct = p \cdot g$
