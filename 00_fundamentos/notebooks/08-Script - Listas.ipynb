# Listas

Una lista en Python esta definida por una secuencia ordenada de elementos que pueden ser dinámicamente alterados.

1. Ordenados significa que cada ítem o elemento tiene un índice basado en su posición.
2. Secuencia significa que los elementos almacenados tienen un orden basado en sus índices.
3. Dinámicamente alterados, significa que cada elemento de la lista puede ser modificado tanto en el índice como el contenido.

# Algunas operaciones básicas con listas

```python
L = [0, 4, 5, -1, 45, 10, 100]
print("El número más pequeño es:", min(L))
print("El número más grande es:", max(L))
L.reverse()
print("La lista en orden inverso es:",L)
L.sort()
print("La lista ordenada de menor a mayor:",L)
L.sort(reverse=True)
print("La lista ordenada de mayor a menor:",L)
```

# Primer script con listas

```python
# Programa que tenga almacenado nombres de paises
# y que muestre los paises que tengan cierto numero de letras

pais = ["México", "España", "Perú", "Argentina", "Chile", "Bolivia", "Ecuador"]

for i in pais:
    if len(i) < 6:
        print(i)
```

# Ahora pedimos al usuario el país que quiere agregar hasta que se canse

```python
# Crear un programa que pida el nombre los paises hasta que escriba 
# la palabra salir y que muestre los paises que cuenten con 
# cierta condición.

paises = []

while True:
    pais = input("Ingresa el nombre de un pais (hasta que escribas salir): ")
    if pais == "salir": 
        break
    paises.append(pais)

for pais in paises:
    if len(pais) < 6:
        print(pais)
```

```python
# Programa que recibe el nombre de un pais y muestra su capital

pais = ["España", "Reino unido", "Irlanda", "Portugal", "Alemania"]
capital = ["Madrid", "Londres", "Dublin", "Lisboa", "Berlin"]

while True:
    nomPais = input("Ingresa el nombre del pais del que quieres saber su capital (Hasta que escribas end): ")
    if nomPais == "end": 
        break
    indice = pais.index(nomPais)
    print("La capital de {} es: {}".format(nomPais, capital[indice]))
```

```python
# Crea un programa que reciba un digito, y que lo escriba en palabras

digito = int(input("Ingresa un digito: "))

letra = ["cero", "uno", "dos", "tres", "cuatro", "cinco", "seis", "siete", "ocho", "nueve"]

print("El digito {} es {}".format(digito, letra[digito]))
```
