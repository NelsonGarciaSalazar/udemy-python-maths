# ¿Qué es un Script?

> En informática, un script, secuencia de comandos o guion​ (traduciendo desde inglés) es un término informal que se usa para designar a un programa relativamente simple. Los scripts regularmente no se compilan con anticipación a código máquina, sino que son ejecutados por un intérprete que lee el archivo de código fuente al momento; o incluso por una consola interactiva donde el usuario suministra el programa al intérprete paso a paso. Los scripts o guiones se pueden usar para prototipar programas, automatizar tareas repetitivas, hacer procesamiento por lotes e interactuar con el sistema operativo y el usuario (debido a esto, los intérpretes de comandos o shells suelen diseñarse con funcionalidades de programación).

# Forma de ejecución de líneas de código

Python es un lenguaje interpretado en el que va ejecutando las líneas de código una a una de manera descendente, por lo que el orden de nuestro programa debe tener una secuencia lógica. Así podemos hacer los siguiente:

````python
print("Esta es la primer línea que se va a interpretar")
print('Esta es la segunda línea del código')
print("Esta es la tercer línea")
````

También podemos hacer *saltos* de línea en donde indicamos donde empieza una línea de código y donde termina con el símbolo de **;**

````python
print("Esta es la primer línea que se va a interpretar"); print('Esta es la segunda línea del código'); print("Esta es la tercer línea")
````

Para escribir cadenas de caracteres por lo regular es utilizar comilla doble ("") pero, también se pueden usar comillas simples (''), además que nos permite hacer ciertas cosas interesantes como escribir:

```
 Hola mundo
 ¡"Hola" desde 'México'!
```



````python
print("Hola mundo")
print('¡"Hola" desde \'México\'!')
````
