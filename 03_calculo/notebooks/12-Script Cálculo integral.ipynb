# Función de probabilidad de densidad

Resuelve la integral
$$\int_{11}^{12}
{\frac{1}{\sqrt{2\pi}}e^{-\frac{(x-10)^2}{2}}}$$

```python
from sympy import *
x = Symbol('x')
p = exp(-(x - 10)**2 / 2) / sqrt(2 * pi)
Integral(p, (x, 11, 12)).doit().evalf()
```

Resuelve la integral
$$\int_{-\infty}^{\infty}
{\frac{1}{\sqrt{2\pi}}e^{-\frac{(x-10)^2}{2}}}$$

```python
Integral(p, (x, S.NegativeInfinity, S.Infinity)).doit().evalf()
```

# Integrales con valor absoluto

Resuelve la integral
$$\int_{-5}^5{|x-5|dx}$$

```python
from sympy import *
x = Symbol('x')
f = abs(x - 5)
Integral(f, (x, -5, 5)).doit()
```

Resuelve la integral
$$\int_0^{x^4}{\cos^2{\theta}d\theta}$$

```python
from sympy import *
x, theta = symbols('x theta')
f = (cos(theta))**2
Integral(f, (theta, 0, x**4)).doit()
```

Determina la derivada de la función
$$g(x) = \int_{2x}^{3x}{\frac{u^2-1}{u^2+1}du}$$

```python
from sympy import *
x, u = symbols('x u')
gu = (u**2 - 1) / (u**2 + 1)
gx = Integral(gu, (u, 2*x, 3*x)).doit()
Derivative(gx, x).doit()
```

# Área bajo la curva

El área $A$ de la región limitada por las curvas $y=f(x),y=g(x)$ y las rectas $x=a,x=b$ donde $f$ y $g$ son continuas y $f(x)\geqslant g(x)$ para toda $x$ en $[a,b]$, es
$$A=\int_a^b{[f(x)-g(x)]dx}$$

```python
from sympy import * 
x = Symbol('x')
f = 5*x - x**2
g = x
Integral(f - g, (x, 0, 4)).doit()
```
