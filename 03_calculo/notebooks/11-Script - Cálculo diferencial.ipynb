# La derivada de una función

Encuentra la derivada de la función

$$S(t) = 5t^2+2t+8$$

```python
from sympy import *
t = Symbol('t')
St = 5*t**2 + 2*t + 8

# Aplicamos la derivada con respecto a t
Derivative(St, t).doit()
```

# Calculadora para derivadas

```python
from sympy import *
from sympy.core.sympify import SympifyError

def derivada(f, var):
    var = Symbol(var)
    d = Derivative(f, var).doit()
    pprint(d)

if __name__ == '__main__':

    f = input('Ingresa la ecuación que quieres derivar: ')
    var = input('Ingresa la variable con la que quieres derivar: ')

    try:
        f = sympify(f)
    except SympifyError:
        print('La función es inválida')
    else:
        derivada(f, var)
```

# Resolviendo máximos y mínimos 

$$x^5-30x^3+50x$$

```python
from sympy import *
x = Symbol('x')
f = x**5 - 30*x**3 + 50*x
d1 = Derivative(f, x).doit()

# Puntos críticos
puntos_criticos = solve(d1)

# Acomodar los puntos criticos
P1 = puntos_criticos[0]
P2 = puntos_criticos[1]
P3 = puntos_criticos[2]
P4 = puntos_criticos[3]

# Segunda derivada de la función
d2 = Derivative(f, x, 2).doit()
```

```python
# Sustituir cada una de las soluciones del punto
# crítico en la segunda derivada
d2.subs({x:P1}).evalf()
```

```python
d2.subs({x:P2}).evalf()
```

```python
d2.subs({x:P3}).evalf()
```

```python
d2.subs({x:P4}).evalf()
```

# Un script que obtiene el valor máximo de una función

```python
from sympy import *

def grad_ascent(x0, f1x, x):
    epsilon = 1e-6
    sept_size = 1e-4
    x_old = x0
    x_new = x_old + sept_size * f1x.subs({x:x_old}).evalf()
    while abs(x_old - x_new) > epsilon:
        x_old = x_new
        x_new = x_old + sept_size * f1x.subs({x:x_old}).evalf()
    
    return x_new

if __name__ == '__main__':

    f = input("Ingresa una función de una variable: ")
    var = input('Ingresa la variable para derivar: ')
    var0 = float(input('Ingresa el valor iniciar de la variable: '))
    try:
        f = sympify(f)
    except SympifyError:
        print('Has ingresado la función incorrectamente')
    else:
        var = Symbol(var)
        d = Derivative(f, var).doit()
        var_max = grad_ascent(var0, d, var)
        print('{0}: {1}'.format(var.name, var_max))
        print('El máximo valor es: {0}'.format(f.subs({var:var_max})))
```
